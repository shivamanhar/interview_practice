```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def bark(self):
        return f"{self.name} saya woof!"
    
my_dog = Dog('Buddy', 3)
print(my_dog.bark())
```

```python
class Cat:
    species = 'Felis catus'
    def __init__(self, name):
        self.name = name
    def meow(self):
        return f"{self.name} saya meow!"
    
    @classmethod
    def get_species(cls, name):
        return cls(f"Baby {name}")
    
```

```python
obj = Cat.get_species("Tom")
```

```python
obj.meow()
```

```python
class Car:
    def __init__(self, name, model):
        self.name = name
        self.model = model
    
    def __repr__(self):
        return f"Car({self.name}, {self.model})"
    
    def __str__(self):
        return f"{self.name} {self.model}"
    
    @classmethod
    def from_string(cls, name, model):
        return cls(name, model)
    
```

```python
obj = Car.from_string("Honda", "Civic")
print(obj.name, obj.model)
print(repr(obj))
print(str(obj))
```
