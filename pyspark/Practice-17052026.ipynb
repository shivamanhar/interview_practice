### Calculate Average Using Window functions

```python
data = [("Alice", 100), ("Bob", 200), ("Cathy", 150), ("David", 300)]
columns = ["Name", "Score"]
df = spark.createDataFrame(data, columns)
```

```python
df.display()
```

```python
window_spec = Window.orderBy('Score').rowsBetween(Window.unboundedPreceding, Window.currentRow)
```

```python
df_with_avg = df.withColumn("Running_avg", F.avg('Score').over(window_spec))
df_with_avg.display()
```

### GroupBy and Aggregations

```python
data = [("HR", 5000), ("HR", 6000), ("IT", 7000), ("IT", 8000), ("Finance", 9000)]
columns = ["Department", "Salary"]
```

```python
df = spark.createDataFrame(data, columns)
```

```python
df.display()
```

```python
agg_df = df.groupBy("Department").agg(F.sum("Salary").alias("Total_Salary"), F.avg("Salary").alias("Avg_Salary"), F.count("salary").alias("Count"))
agg_df.display()
```

### Rank Data Using Window Functions

```python
data = [("Alice", 85), ("Bob", 78), ("Cathy", 85), ("David", 90)]
columns = ["Name", "Score"]
df = spark.createDataFrame(data, columns)
# Define a window
window_spec = Window.orderBy(df["Score"].desc())

# Rank the rows
df_with_rank = df.withColumn("Rank", F.rank().over(window_spec))
df_with_rank.display()
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

window_spec = Window.orderBy(F.col("Score").desc())

df_with_rank = df.withColumn(
    "Rank",
    F.rank().over(window_spec)
)

df_with_rank.display()
```

```python
data = [("Alice", 85), ("Bob", 78), ("Cathy", 85), ("David", 90)]
columns = ["Name", "Score"]
df = spark.createDataFrame(data, columns)
```

```python
df = df.withColumn('n',F.lit(1))
```

```python
window_spec = Window.partitionBy('n').orderBy(F.col("Score").desc())
```

```python
df_with_rank  = df.withColumn("Rank", F.rank().over(window_spec))
```

```python
df_with_rank.display()
```

```python
df_with_rank  = df.withColumn("Rank", F.dense_rank().over(window_spec))
```

```python
df_with_rank.display()
```

### Explode an Array Column

### Calculate Moving Average

```python
data = [("2023-01-01", 100), ("2023-01-02", 200), ("2023-01-03", 300), ("2023-01-04", 400)]
columns = ["Date", "Value"]

df = spark.createDataFrame(data, columns)

window_spec = Window.orderBy("Date").rowsBetween(-1, 1)

df_with_moving_avg = df.withColumn("Moving_Avg", F.avg("Value").over(window_spec))
df_with_moving_avg.display()
```
