```python
# 1. Find duplicate elements in a list
def find_duplicates(nums):
    seen = set()
    duplicates = set()
    for num in nums:
        if num in seen:
            duplicates.add(num)
        else:
            seen.add(num)
    return list(duplicates)

nums = [1, 2, 3, 4, 5, 1, 2]
print(find_duplicates(nums))
```

```python
# 2. Count word frequency

def word_count(sentence):
    words = sentence.lower().split()
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0)+1
    return freq

# Example
text = "data engineering data pipeline"
print(word_count(text))
```

```python
#5. Find the missing number from 1 to n
def missing_number(nums, n):
    expected_sum = n*(n+1)//2
    actual =sum(nums)
    return expected_sum - actual

nums = [1, 2, 3, 5]
print(missing_number(nums, 5))
```

```python
#6. Flatten a nested list.

def flatten(lst):
    result = []
    for item in lst:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result

nested_list = [1, [2, [3, 4], 5], 6]
print(flatten(nested_list))
```

```python
#6. Remove duplicates while preserving order
def remove_duplicates(items):
    seen = set()
    result = []
    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result

print(remove_duplicates([1, 2, 3, 2, 4, 1]))
```

```python
#8. Group Data by key
employees = [
    {"name": "Alice", "dept": "IT"},
    {"name":"Bob", "dept":"HR"},
    {"name":"Charlie", "dept":"IT"}
]
grouped = {}

for emp in employees:
    dept = emp["dept"]
    if dept not in grouped:
        grouped[dept] = []
    grouped[dept].append(emp['name'])

print(grouped)
```

```python
#9. Create a generator for streaming data
def read_numbers(n):
    for i in range(n):
        yield i

for num in read_numbers(5):
    print(num)
```

```python
#14. Check Balanced Parentheses

def is_balanced(s):
    stack = []

    for char in s:
        if char == "(":
            stack.append(char)
        elif char == ")":
            if not stack:
                return False
            stack.pop()

    return len(stack) == 0

print(is_balanced("((12(h)))"))
print(is_balanced("(()))"))
#15. Find the first non-repeated character in a string
```

```python
#15. ETL Transformation Example
sales = [
    {'product':'A', "price": 100},
    {'product':'B', 'price': 200}
]

transformed = []

for item in sales:
    transformed.append({
        'product': item['product'],
        'price': item['price'] * 1.18
    })

print(transformed)
```
