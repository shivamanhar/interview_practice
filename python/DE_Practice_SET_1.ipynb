```python
# 1. Find duplicate elements in a list
def find_duplicates(nums):
    seen = set()
    duplicates = set()
    for num in nums:
        if num in seen:
            duplicates.add(num)
        else:
            seen.add(num)
    return list(duplicates)

nums = [1, 2, 3, 4, 5, 1, 2]
print(find_duplicates(nums))
```

```python
# 2. Count word frequency

def word_count(sentence):
    words = sentence.lower().split()
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0)+1
    return freq

# Example
text = "data engineering data pipeline"
print(word_count(text))
```

```python
# 3. Remove Null values from dataset
data = [
    {"id": 1, "name":"alice"},
    {"id": 2, "name": None},
    {"id": 3, "name":"Bob"}
]

print(data)

cleaned = [row for row in data if row["name"] is not None]
print(cleaned)
```
