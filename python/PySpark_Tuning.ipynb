```python
df = spark.read.option("header", "true").csv("/Volumes/workspace/sh20508860/raw/order_details/Order Details.csv")
```

```python
df.display()
```

```python
from pyspark.sql import functions as F
cloth_df = df.groupBy('Category').agg(F.sum('Amount').alias('Amount'))
```

```python
cloth_df.display()
```

```python
from pyspark.sql import Window

w = Window.partitionBy('Category').orderBy(F.col('Amount').asc())
final_df = df.withColumn('row_number', F.row_number().over(w))
final_df.display()
```
