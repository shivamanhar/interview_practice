```python
# Remove Duplicates from a DataFrame

data  =[("Jonh", 28, "M"),("Alice", 25, "F"),("Bob", 30, "M"),("Alice", 25, "F"), ("Bob", 30, "M")]

columns = ["Name", "Age", "Gender"]

df = spark.createDataFrame(data, columns)
df.display()
```

```python
unique_data = df.dropDuplicates(['Gender'])

unique_data.display()
```

```python
df.display()
```

```python
without_duplicates_df = df.dropDuplicates()
without_duplicates_df.display()
```

```python
# Daily user retention for a 30 day cohort window 
data = [(1, '2023-01-01'), (1, '2023-01-02'), (2, '2023-01-01'), (2, '2023-01-03'), (3, '2023-01-02')]

df = spark.createDataFrame(data, ['user_id','login_date'])
df.printSchema()
```

```python
from pyspark.sql import functions as F
```

```python
df = df.withColumn('login_date', F.to_date(F.col('login_date')))
df.printSchema()
```

```python
df.display()
```

```python
cohort = df.groupBy('user_id').agg(F.min('login_date').alias('cohort_date'))
```

```python
cohort.display()
```

```python
df = df.join(cohort, on="user_id")
df.display()
```

```python
df = df.withColumn('days_since_signup', F.datediff('login_date', 'cohort_date'))
df.display()
```

```python
events = [(1, 'login', '2023-01-01 10:00'), (1, 'click', '2023-01-01 10:05'), (1, 'logout', '2023-01-01 10:10'),
          (1, 'login', '2023-01-02 10:00'), (2, 'login', '2023-01-01 09:00')]

df = spark.createDataFrame(events, ["user_id", "event", "ts"])
df.display()
```

```python
df = df.withColumn('ts', F.to_timestamp("ts"))
```

```python
df.display()
```

```python
from pyspark.sql import Window
```

```python
w = Window.partitionBy("user_id").orderBy(F.col('ts').desc())
```

```python
df = df.withColumn("rn", F.row_number().over(w))
```

```python
filter_df = df.where(F.col("rn") <= 3)
filter_df.display()
```
