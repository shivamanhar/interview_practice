```python
def word_count(sentence):
    words = sentence.lower().split()
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0)+1
    return freq

text = "data engineering data pipeline data engineering data pipeline"
print(word_count(text))
```

```python
def char_count(word):
    chars = list(word)
    freq = {}
    for c in chars:
        freq[c] = freq.get(c, 0)+1
    
    return freq

print(char_count("shiva manhar"))
```

```python
def missing_number(nums, n):
    expeacted=  n*(n+1)//2
    actual = sum(nums)
    return expeacted - actual

nums = [1, 2, 3, 5]
print(missing_number(nums, 5))
```

```python
def flatten(lst):
    result = []

    for item in lst:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result

lst = [1, 2, [3, 4], [5, [6, 7]]]
print(flatten(lst))
```

```python
# SecondLargest
def getSecondLargest(self, arr):
        # code here
        largest = -1
        second = -1

        for num in arr:
            if num > largest:
                second = largest
                largest = num
            elif num != largest and num > second:
                second = num

        return second
```

```python
arr = [2, 5, 6, 9, 5, 6, 2]
lenght  = len(arr)
darr = []

for i in range(lenght):
    for j in range(i+1, lenght):
        if arr[i] == arr[j]:
            darr.append(arr[i])
print(darr)
```

```python
arr = [6, 9, 7, 3, 2, 6, 9, 3]

dup = []
see = set()

for i in arr:
    if i not in see:
        see.add(i)
    else:
        dup.append(i)

print(dup)
```
